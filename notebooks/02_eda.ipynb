```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file_path = "../data/Palo Alto Networks.xlsx"

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print(df.shape)
```

```python
attrition_counts = df["Attrition"].value_counts()

print(attrition_counts)

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Attrition")
plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.show()
```

```python
attrition_percentage = df["Attrition"].value_counts(normalize=True) * 100

print(attrition_percentage.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Department",
    hue="Attrition"
)

plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=20)
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="OverTime",
    hue="Attrition"
)

plt.title("Employee Attrition by Overtime")
plt.xlabel("OverTime")
plt.ylabel("Number of Employees")
plt.show()
```

```python
overtime_attrition = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100

print(overtime_attrition.round(2))
```

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    y="JobRole",
    hue="Attrition"
)

plt.title("Employee Attrition by Job Role")
plt.xlabel("Number of Employees")
plt.ylabel("Job Role")
plt.show()
```
